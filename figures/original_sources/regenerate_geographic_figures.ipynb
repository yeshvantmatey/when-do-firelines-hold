# Rebuild the manuscript maps

Use the project `.venv` Python kernel in Antigravity. The notebook reads the saved study data and downloaded Census boundaries. It rebuilds the geographic coverage map and the observed NIFC–NIROPS example. No model is fitted. Original data and manuscript files remain unchanged. Only the new `outputs/manuscript_maps_20261110` folder is updated.


```python
from pathlib import Path
import subprocess
import sys
import shutil
from IPython.display import Image, display

start = Path.cwd().resolve()
repo = next((p for p in [start, *start.parents] if (p / 'notebooks').is_dir() and (p / 'data/processed').is_dir()), None)
if repo is None:
    raise RuntimeError('Open the notebook from the wildfire_fireline_effectiveness workspace.')
out = repo / 'outputs/manuscript_maps_20261110'
print('Python:', sys.executable)
print('Workspace:', repo)
```

## Geographic coverage
The map verifies 381 fires and 304,143 retained fireline units. Each point is the mean longitude and latitude of a fire's retained units. The star identifies the fire used for the local example.


```python
subprocess.run([sys.executable, str(out / 'study_extent/build_study_extent.py')], cwd=repo, check=True)
display(Image(filename=str(out / 'study_extent/fig_study_extent.png'), width=850))
```

## Observed NIFC–NIROPS example
The helper verifies source labels, retained membership, prediction time, checked positions, crossing evidence, and the Held follow-up sequence. Recorded NIROPS times approximate imagery collection times. The display uses vector perimeters; original selection checks used a 25 m grid.


```python
subprocess.run([sys.executable, str(out / 'nifc_nirops_example/build_nifc_nirops_example.py')], cwd=repo, check=True)
display(Image(filename=str(out / 'nifc_nirops_example/nifc_nirops_observed_example.png'), width=1200))
```

## Refresh the Overleaf assets
The baseline figure is copied unchanged from the existing supplement. The package includes insertion snippets and panel-reference corrections. Rendering the main manuscript remains a separate step.


```python
target = out / 'overleaf_upload/editorial_figures'
target.mkdir(parents=True, exist_ok=True)
for ext in ['pdf', 'png']:
    shutil.copy2(out / f'study_extent/fig_study_extent.{ext}', target / f'fig_study_extent.{ext}')
    shutil.copy2(out / f'nifc_nirops_example/nifc_nirops_observed_example.{ext}', target / f'fig_nifc_nirops_example.{ext}')
shutil.copy2(repo / 'overleaf_revision_08_series_20261001/combined_supplement_final/supplement_08e_figures/08e_forward_comparisons.pdf', target / 'fig_baseline_comparison.pdf')
archive = shutil.make_archive(str(out / 'fireline_figures_overleaf'), 'zip', out / 'overleaf_upload')
print(archive)
```
