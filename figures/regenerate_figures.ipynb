# Reproduce the added manuscript figures
Run from the code release or its figures folder. Set OUTPUT_ROOT to a new directory outside the three release folders. No model is fitted; the packaged saved data are read-only.
The historical editorial redraw is optional and is not the current pooled-results figure.

```python
from pathlib import Path
import subprocess, sys
candidates = [Path.cwd(), *Path.cwd().parents]
FIGURES = next(p for p in [*(q / "figures" for q in candidates), *candidates] if (p / "figure_common.py").is_file())
OUTPUT_ROOT = FIGURES.resolve().parents[2] / "reproduced_fireline_figures"
assert not OUTPUT_ROOT.exists(), "Choose a new output directory."
OUTPUT_ROOT.mkdir(parents=True)
print("Using packaged figure inputs; no training.")
```

```python
for name in ["study_extent", "nifc_nirops_example", "schematic"]:
    subprocess.run([sys.executable, str(FIGURES / f"reproduce_{name}.py"), "--output", str(OUTPUT_ROOT / name)], check=True)
```
