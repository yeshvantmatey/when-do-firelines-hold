# Complete the three missing Table II bootstrap intervals

No models are fitted. The calculation uses the saved primary LightGBM out-of-fold predictions for 304,143 fireline units from 381 fires.

Each of 5,000 replicates samples 381 complete fires with replacement. Each selected fire contributes every unit, repeated according to the fire multiplicity. Pooled metrics continue to give units equal weight within the resampled dataset. The sorted fire order, NumPy generator, seed (`20260825`), and linear percentile intervals match notebook 05E. The code reproduces all six existing intervals and fills only the three missing intervals (mean predicted Held probability, pooled AP, pooled ROC-AUC).

The implementation preserves probability ties and verifies its weighted AP and ROC-AUC against scikit-learn, including a literal duplicate-fire check. It hashes input files before and after calculation. New timestamped output folders preserve previous results.


```python
from pathlib import Path
import importlib.util

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "scripts" / "complete_table_ii_bootstrap.py").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
SCRIPT = PROJECT_ROOT / "scripts" / "complete_table_ii_bootstrap.py"
assert SCRIPT.exists(), "Run the notebook from the project root or notebooks directory."
spec = importlib.util.spec_from_file_location("table_ii_bootstrap", SCRIPT)
analysis = importlib.util.module_from_spec(spec)
spec.loader.exec_module(analysis)
```

```python
# Execute the fixed-prediction analysis. No model training occurs.
run_directory, table_ii = analysis.run()
display(table_ii)
```
